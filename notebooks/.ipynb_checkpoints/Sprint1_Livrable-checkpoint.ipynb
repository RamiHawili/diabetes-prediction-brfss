## 1. Contexte

&nbsp;&nbsp;&nbsp;&nbsp;Aux États-Unis, le diabète est l'une des maladies chroniques les plus répandues. En 2018, les CDC (les Centres pour le contrôle et la prévention des maladies) estimaient à 34,2 millions le nombre d'Américains diabétiques, et à 88 millions ceux en état de prédiabète; dont une grande majorité ignorait leur situation.

Cette maladie cause un dérèglement du sucre sanguin, ce qui, sur le long terme, peut engendrer des complications graves : maladies cardiovasculaires, insuffisance rénale, amputations, perte de la vue. Son coût économique est estimé à près de 400 milliards de dollars par an.

Le dataset utilisé provient du *Behavioral Risk Factor Surveillance System (BRFSS)*, une enquête téléphonique annuelle menée par les CDC auprès de plus de 400 000 Américains. Notre version (2015) contient 253 680 réponses et 22 variables relatives aux comportements de santé, antécédents médicaux et conditions socio-économiques des participants.


&nbsp;&nbsp;&nbsp;&nbsp;Le **livrable 1** constitue la partie "prétraitement" du projet. C'est dans ce dernier que le dataset issu de l'étude BRFSS va être analysé, normalisé et transformé pour que par la suite il soit plus facile d'utiliser ses données à des fins d'entraînement de modèles d'IA de Deep Learning et d'analyses statistiques.  
L'échéance de ce livrable est fixée au 15 mai 2026 à 8h.

## 2. Objectif

Construire un modèle de classification binaire capable de prédire si une personne est diabétique à partir de ses indicateurs de santé.

La variable cible `Diabetes_012` est binarisée comme suit :
- **0** → pas de diabète ou prédiabète (classes 0 et 1 regroupées)
- **1** → diabète (classe 2)

La métrique finale à maximiser est la **ROC AUC** (mesure la capacité du modèle à séparer les classes positives des négatives) sur un jeu de test non étiqueté.

## 3. Problématique

> *À partir des indicateurs de santé comportementaux et médicaux recueillis dans le BRFSS 2015, est-il possible de correctement prédire si une personne souffre de diabète ?*

Ce sprint se concentre uniquement sur la préparation des données : nettoyage, exploration, et normalisation. Les modèles seront construits dans les sprints suivants.

## 4. Contraintes

- Les données provenant du BRFSS présentent un fort déséquilibre entre les classes, puisque regrouper "no diabetes" et "prediabetes" en une seule classe rend la classe minoritaire (diabète) bien moins représentée. C'est pourquoi la métrique retenue est la ROC AUC plutôt que l'accuracy, qui masquerait ce problème.
  
- Le dataset étant issu d'une enquête téléphonique déclarative, les réponses dépendent entièrement de ce que les participants ont bien voulu déclarer, ce qui peut introduire des biais qu'on ne peut pas corriger.  
- Les 22 variables du dataset sont de nature mixte, certaines sont binaires, d'autres ordinales ou continues. Chaque type nécessite un traitement différent lors du prétraitement et de la normalisation, ce qui rend le pipeline plus complexe.  
- Avec 253 680 répondants, le volume de données est conséquent. Il faut faire attention à la performance de calcul et la gestion mémoire tout au long du projet.  
- Enfin, le dataset contient des variables sensibles comme la tension artérielle, le cholestérol ou le sex. Une analyse RGPD et éthique est nécessaire pour justifier les choix de conservation ou de suppression de certaines variables pour éviter les discriminations et les mauvaises interprétations.

## 5. Le dataset et ses variables

L'enquête BRFSS nous a permis de construire un tableau de données de 253 680 lignes et 22 colonnes, toutes encodées en numérique.

**Détaillons ici chaque variable pour en comprendre le sens :**

- **Diabetes_012** (cible, 0/1/2) : indique si la personne n'a pas de diabète (0), est prédiabétique (1) ou diabétique (2)
- **HighBP** (binaire, 0/1) : indique si la personne souffre d'hypertension artérielle
- **HighChol** (binaire, 0/1) : indique si la personne a un taux de cholestérol élevé
- **CholCheck** (binaire, 0/1) : indique si la personne a fait un contrôle du cholestérol au cours des 5 dernières années
- **BMI** (valeur numérique) : indice de masse corporelle de la personne
- **Smoker** (binaire, 0/1) : indique si la personne a fumé au moins 100 cigarettes dans sa vie, soit l'équivalent de 5 paquets
- **Stroke** (binaire, 0/1) : indique si la personne a déjà eu un AVC déclaré par un médecin
- **HeartDiseaseorAttack** (binaire, 0/1) : indique si la personne a déjà eu une maladie coronarienne ou un infarctus du myocarde
- **PhysActivity** (binaire, 0/1) : indique si la personne a pratiqué une activité physique dans les 30 derniers jours, hors activité professionnelle
- **Fruits** (binaire, 0/1) : indique si la personne consomme des fruits au moins une fois par jour
- **Veggies** (binaire, 0/1) : indique si la personne consomme des légumes au moins une fois par jour
- **HvyAlcoholConsump** (binaire, 0/1) : indique si la personne a une consommation excessive d'alcool
- **AnyHealthcare** (binaire, 0/1) : indique si la personne dispose d'une couverture santé
- **NoDocbcCost** (binaire, 0/1) : indique si la personne a déjà renoncé à consulter un médecin pour des raisons financières
- **GenHlth** (ordinale) : état de santé général auto-évalué par la personne (1 = excellent, 5 = mauvais)
- **MentHlth** (continue) : nombre de jours de mauvaise santé mentale ressentis sur les 30 derniers jours
- **PhysHlth** (continue) : nombre de jours de mauvaise santé physique ressentis sur les 30 derniers jours
- **DiffWalk** (binaire, 0/1) : indique si la personne a des difficultés à marcher ou monter des escaliers
- **Sex** (binaire, 0/1) : indique le sexe de la personne (0 = femme, 1 = homme)
- **Age** (ordinale) : tranche d'âge de la personne
- **Education** (ordinale) : niveau d'études de la personne
- **Income** (ordinale) : tranche de revenus de la personne

## 6. Analyse éthique des variables à retirer de l'étude

&nbsp;&nbsp;&nbsp;&nbsp;Dans le domaine de la data et de l'intelligence artificielle (IA), il faut constamment questionner l'éthique derrière les données qui vont être utilisées dans des études.   
Que ce soit de l'analyse de données ou de l'entraînement d'IA, il est important d'assurer l'anonymat des individus et de prévenir les dérives d'analyse dues à tout type de discrimination.   
C'est dans cet objectif que nous avons décidé de retirer quelques variables du dataset, que ce soit parce qu'elles n'ont aucune utilité vis-à-vis de l'étude menée ou parce que l'éthique et la protection des données (RGPD) le demandent.  

- La variable **"Sex"** : Médicalement le sexe peut avoir une légère influence sur le diabète de type 2, mais étant donné que nous prédisons à partir de comportements et d'antécédents médicaux et non de caractéristiques biologiques, il n'est pas utile de la garder. De plus, intégrer le sexe comme variable prédictive pourrait amener le modèle à donner des résultats différents selon le genre ce qui constitue une discrimination de genre qu'il faut éviter.
  
- La variable **"Income"** : Le revenu actuel d'une personne n'est pas un indicateur médical du diabète. Se baser dessus pour prédire une maladie pourrait pénaliser les personnes à faibles revenus qui n'ont pas forcément plus de risques biologiques que les autres, mais simplement moins accès aux soins. Cela amènerait le modèle à des conclusions incorrectes et discriminatoires qu'il faut éviter.

  
- La variable **"Education"** : Le niveau d'études n'est pas un facteur clinique lié au diabète, c'est un facteur social. Un médecin ne diagnostique pas le diabète en fonction du diplôme de son patient. Conserver cette variable amènerait le modèle à discriminer selon le niveau d'études, ce qui n'est ni justifié médicalement ni acceptable éthiquement.

## 10. Étape 2 : Typage des variables

On distingue trois catégories de variables :

- **Variables binaires (0/1)** : les 13 variables indicatrices (HighBP, Smoker, Stroke, etc.) qui prennent uniquement la valeur 0 ou 1. Elles sont conservées en l'état en int8.
  
- **Variables ordinales** : GenHlth et Age. Ce sont des catégories ordonnées (ex. GenHlth va de 1 = excellent à 5 = mauvais, Age correspond à des tranches d'âge numérotées). On sait que 2 > 1, mais on ne peut pas dire que la différence entre 1 et 2 est la même qu'entre 4 et 5.

- **Variables continues** : BMI, MentHlth, PhysHlth. Ce sont des mesures numériques réelles où l'écart entre deux valeurs est interprétable (ex. un BMI de 30 est objectivement le double de 15). Elles seront standardisées et typées en float32.

```python
# Typage explicite
BINARY_COLS = [
    "HighBP", "HighChol", "CholCheck", "Smoker", "Stroke",
    "HeartDiseaseorAttack", "PhysActivity", "Fruits", "Veggies",
    "HvyAlcoholConsump", "AnyHealthcare", "NoDocbcCost", "DiffWalk"
]
ORDINAL_COLS = ["GenHlth", "Age"]
NUMERIC_COLS = ["BMI", "MentHlth", "PhysHlth"]

# Application sur df_raw
df_raw[BINARY_COLS]       = df_raw[BINARY_COLS].astype("int8")
df_raw[ORDINAL_COLS]      = df_raw[ORDINAL_COLS].astype("category")
df_raw[NUMERIC_COLS]      = df_raw[NUMERIC_COLS].astype("float32")

print("Variables binaires :", BINARY_COLS)
print()
print("Variables ordinales :", ORDINAL_COLS)
print()
print("Variables numériques:", NUMERIC_COLS)
print()
print(df_raw.dtypes)
print(f"\nMémoire estimée (df_raw) : {df_raw.memory_usage(deep=True).sum() / 1e6:.2f} MB")
```

## 11. Étape 3 : Nettoyage des doublons et gestion des valeurs manquantes

### 11.1 Valeurs manquantes

```python
# Valeurs manquantes
n_missing = df_raw.isnull().sum().sum()
print(f"Total valeurs manquantes : {n_missing}")

if n_missing == 0:
    print("Aucun traitement nécessaire. Le BRFSS ne contient pas de NA dans cette version.")
else:
    print(f" {n_missing} valeurs manquantes détectées, application du protocole de nettoyage...")
    
    for col in BINARY_COLS:
        if df[col].isnull().sum() > 0:
            mode = df[col].mode()[0]
            df[col] = df[col].fillna(mode)
            print(f"   {col} : NA remplacés par le mode ({mode})")
    
    for col in NUMERIC_COLS:
        if df[col].isnull().sum() > 0:
            mediane = df[col].median()
            df[col] = df[col].fillna(mediane)
            print(f"   {col} : NA remplacés par la médiane ({mediane})")
    
    print(f"Valeurs manquantes après traitement : {df.isnull().sum().sum()}")
```

### 11.2 Doublons

Un doublon exact correspond à deux répondants ayant exactement le même profil sur les 21 variables. Étant donné le nombre important de variables binaires, de tels doublons sont plausibles statistiquement et on les supprime.

```python
n_before  = len(df_raw)
n_dup     = df_raw.duplicated().sum()

df_clean  = df_raw.drop_duplicates().reset_index(drop=True)
n_after   = len(df_clean)

print(f"Doublons détectés : {n_dup:,}")
print(f"Lignes avant : {n_before:,} & après : {n_after:,} (supprimés : {n_before - n_after:,})")

print("\nDistribution cible après nettoyage :")
print(df_clean["Diabetes_012"].value_counts().sort_index())
```

## 12. Étape 4 : Séparation cible / variables explicatives

On isole la variable cible `Diabetes_012` et on la binarise : les classes 0 (pas de diabète) et 1 (prédiabète) sont regroupées sous la valeur 0. La classe 2 (diabète avéré) devient 1.

```python
# Distribution initiale de la cible
print("Distribution Diabetes_012 avant binarisation :")
print(df_raw["Diabetes_012"].value_counts().sort_index())

# Binarisation
df = df_raw.copy()
df["Diabetes_binary"] = (df["Diabetes_012"] == 2).astype(int)
df = df.drop(columns=["Diabetes_012"])

# Typage de la variable cible maintenant qu'elle existe
df["Diabetes_binary"] = df["Diabetes_binary"].astype("int8")

print("\nDistribution après binarisation :")
print(df["Diabetes_binary"].value_counts().sort_index())
print(f"Taux positif : {df['Diabetes_binary'].mean():.3%}")
```

```python
# Séparation X / y
TARGET = "Diabetes_binary"
FEATURES = []
for c in df.columns:
    if c != TARGET:
        FEATURES.append(c)

X_full = df[FEATURES]
y_full = df[TARGET]

print(f"Variables explicatives : {len(FEATURES)}")
print(f"Variable cible : {TARGET}")
print(f"Shape X : {X_full.shape}  &&  Shape y : {y_full.shape}")
```

## 13. Étape 5 : Analyse exploratoire des données (EDA)

Avant de construire le modèle, on explore les données pour répondre à des questions concrètes :  
- le dataset est-il déséquilibré ?  
- Quelles variables sont les plus liées au diabète ?  
- Y a-t-il des colinéarités entre les variables ?
  
Cette étape guide les choix de modélisation du sprints suivants.

```python
# 1. Distribution de la variable cible
counts         = df["Diabetes_binary"].value_counts().sort_index()
ratios         = counts / counts.sum()
imbalance_ratio = counts[0] / counts[1]

print("Effectifs :")
print(f"  Classe 0 (pas diabétique) : {counts[0]:,}  ({ratios[0]:.2%})")
print(f"  Classe 1 (diabétique)     : {counts[1]:,}  ({ratios[1]:.2%})")
print(f"  Imbalance ratio           : {imbalance_ratio:.2f} pour 1")

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(["Classe 0\n(pas diabétique)", "Classe 1\n(diabétique)"],
       counts.values, color=["steelblue", "tomato"])
for i, (val, ratio) in enumerate(zip(counts.values, ratios.values)):
    ax.text(i, val + 500, f"{val:,}\n({ratio:.1%})", ha="center", fontsize=10)
ax.set_title("Distribution de la variable cible Diabetes_binary")
ax.set_ylabel("Nombre d'individus")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "class_distribution.png", dpi=120)
plt.show()
```

### Analyse

- Le dataset est fortement déséquilibré : 86.1% des individus n'ont pas de diabète contre 13.9% qui en ont.   
- Ce déséquilibre est important à prendre en compte car un modèle mal configuré pourrait apprendre à toujours prédire "pas diabétique" et sembler performant sans jamais détecter un seul cas réel.  
- C'est pourquoi la métrique retenue est la ROC AUC plutôt que l'accuracy.

```python
# 2. Quelles variables sont les plus liées au diabète ?
corr        = df.drop(columns=["Diabetes_binary"]).corrwith(df["Diabetes_binary"])
corr_sorted = corr.abs().sort_values(ascending=False)

print("Top 10 corrélations (valeur absolue) avec Diabetes_binary :")
for feat, val in zip(corr_sorted.index[:10], corr.loc[corr_sorted.index[:10]]):
    print(f"  {feat:<25} {val:+.3f} ")
```

### Analyse

- Les 3 variables les plus corrélées avec le diabète sont GenHlth (+0.294), HighBP (+0.263) et DiffWalk (+0.218).
- La seule corrélation négative est PhysActivity (-0.118) qui dit que plus une personne est active physiquement, moins elle a de risque d'être diabétique.
- La variable Stroke est la moins corrélée parmi les positives (+0.106), ce qui suggère qu'un AVC est davantage une conséquence du diabète qu'un facteur prédictif direct.

```python
# 3. Taux de diabète selon chaque facteur de risque binaire
taux = {}
for col in BINARY_COLS:
    taux[col] = df[df[col] == 1]["Diabetes_binary"].mean() * 100

taux_series = pd.Series(taux).sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(taux_series.index, taux_series.values, color="steelblue")
ax.axvline(df["Diabetes_binary"].mean() * 100, color="tomato",
           linestyle="--", linewidth=1.5,
           label=f"Taux global ({df['Diabetes_binary'].mean()*100:.1f}%)")
for bar, val in zip(bars, taux_series.values):
    ax.text(val + 0.2, bar.get_y() + bar.get_height() / 2,
            f"{val:.1f}%", va="center", fontsize=9)
ax.set_title("Taux de diabète parmi les individus présentant chaque facteur de risque (variable = 1)")
ax.set_xlabel("Taux de diabète (%)")
ax.legend()
plt.tight_layout()
plt.savefig(FIGURES_DIR / "diabetes_rate_by_risk_factor.png", dpi=120)
plt.show()
```

### Analyse

- Les individus avec une maladie cardiaque (HeartDiseaseorAttack) présentent le taux de diabète le plus élevé à 33.0%, suivis de ceux ayant eu un AVC (Stroke, 31.8%) et de ceux ayant des difficultés à marcher (DiffWalk, 30.7%), etc. Cela montre qu'il sont des facteurs de risques réelles. 
- À l'inverse, certaines variables comme "HvyAlcoholConsump" notamment sont en dessous du taux global, ce qui est parfois contre-intuitif mais peut s'expliquer. Par exemple pour la variable "HvyAlcoholConsump", les gros consommateurs d'alcool sont en moyenne plus jeunes et en meilleure santé physique dans ce dataset.

```python
# 4. Taux de diabète par tranche d'âge
age_labels = {
    1: "18-24", 2: "25-29", 3: "30-34", 4: "35-39", 5: "40-44",
    6: "45-49", 7: "50-54", 8: "55-59", 9: "60-64", 10: "65-69",
    11: "70-74", 12: "75-79", 13: "80+"
}

taux_age = (
    df.groupby("Age")["Diabetes_binary"]
    .mean() * 100
).rename(index=age_labels)

fig, ax = plt.subplots(figsize=(12, 5))
bars = ax.bar(taux_age.index, taux_age.values, color="steelblue")
ax.axhline(df["Diabetes_binary"].mean() * 100, color="tomato",
           linestyle="--", linewidth=1.5,
           label=f"Taux global ({df['Diabetes_binary'].mean()*100:.1f}%)")
for bar, val in zip(bars, taux_age.values):
    ax.text(bar.get_x() + bar.get_width() / 2, val + 0.3,
            f"{val:.1f}%", ha="center", fontsize=8)
ax.set_title("Taux de diabète par tranche d'âge")
ax.set_xlabel("Tranche d'âge")
ax.set_ylabel("Taux de diabète (%)")
ax.legend()
plt.tight_layout()
plt.savefig(FIGURES_DIR / "diabetes_rate_by_age.png", dpi=120)
plt.show()
```

### Analyse

## 14. Étape 6 : Découpage train / validation / test

On effectue le découpage avant le nettoyage approfondi pour éviter tout data leakage (les statistiques de normalisation ne doivent pas "voir" le test).

Le découpage est stratifié sur la variable cible pour conserver le même ratio de positifs dans chaque partition.

- **Train** : 70%  
- **Validation** : 15%  
- **Test** : 15%

```python
# Découpage du dataset en 3 parties : Train / Validation / Test

# Premier split : on sépare 70% pour l'entraînement et 30% temporaire
# Le "temp" contiendra la validation et le test (qu'on séparera ensuite en 2)
X_train_raw, X_temp, y_train_raw, y_temp = train_test_split(
    X_full,              # les features (variables explicatives)
    y_full,              # la cible (0 = pas diabétique, 1 = diabétique)
    test_size=0.30,      # 30% des données vont dans le set temporaire
    stratify=y_full,     # conserve le même ratio de positifs (15%) dans chaque partie
    random_state=RANDOM_STATE  # fixe la graine aléatoire pour reproduire les mêmes résultats
)

# Deuxième split : on coupe le "temp" en deux parties égales (15% / 15%)
# test_size=0.50 signifie 50% du temp pour val et 50% pour test
# comme le temp = 30% du total, on obtient bien 15% val et 15% test
X_val_raw, X_test_raw, y_val, y_test = train_test_split(
    X_temp,              # le set temporaire (30% du dataset)
    y_temp,              # les cibles correspondantes
    test_size=0.50,      # 50% du temp → validation | 50% du temp en test
    stratify=y_temp,     # conserve le même ratio de positifs dans val et test
    random_state=RANDOM_STATE
)

# Vérification : affichage du nombre de lignes et du taux de positifs
# On vérifie que chaque split a bien ~15.27% de positifs (stratification réussie)
for name, X, y in [
    ("Train", X_train_raw, y_train_raw),
    ("Val  ", X_val_raw,   y_val),
    ("Test ", X_test_raw,  y_test)
]:
    print(f"{name} : {len(X):>7,} lignes  &&  taux positif : {y.mean():.3%}")
```

**Note RGPD/Éthique :**  
La suppression des doublons exacts est une mesure de minimisation des données conforme à l'art. 5.1.c du RGPD. Conserver des données en double sans raison va à l'encontre de ce principe.  

Cela évite également de surestimer le poids de certains profils : si un même profil apparaît plusieurs fois, le modèle lui accorderait plus d'importance qu'il ne le mérite, ce qui ne serait pas représentatif de la réalité.

## 15. Étape 7 : Rééquilibrage du dataset avec SMOTE

Le dataset présente un fort déséquilibre entre les classes : environ 86% de non-diabétiques contre 14% de diabétiques. Un modèle entraîné sur ces données aurait tendance à systématiquement prédire la classe majoritaire, ce qui serait catastrophique dans un contexte médical.

On applique **SMOTE** (Synthetic Minority Oversampling Technique) pour rééquilibrer les classes à 50/50 uniquement sur X_train. SMOTE génère des exemples synthétiques de la classe minoritaire en interpolant entre des points existants (il ne duplique pas simplement des lignes).

X_val et X_test restent avec leur distribution naturelle car ils doivent représenter la réalité du terrain.

```python
# Affichage AVANT rééquilibrage
print("Distribution AVANT SMOTE :")
print(f"  Non diabétique (0) : {(y_train_raw == 0).sum()}")
print(f"  Diabétique     (1) : {(y_train_raw == 1).sum()}")
print(f"  Ratio : {(y_train_raw == 0).sum() / (y_train_raw == 1).sum():.2f}")

# Application de SMOTE sur X_train / y_train uniquement
# sampling_strategy=1.0 -> on vise du 50/50
smote = SMOTE(sampling_strategy=1.0, random_state=42)
X_train, y_train = smote.fit_resample(X_train_raw, y_train_raw)

# Affichage APRÈS rééquilibrage
print("\nDistribution APRÈS SMOTE :")
print(f"  Non diabétique (0) : {(y_train == 0).sum()}")
print(f"  Diabétique     (1) : {(y_train == 1).sum()}")
print(f"  Ratio : {(y_train == 0).sum() / (y_train == 1).sum():.2f}")
```

```python
counts_apres = [(y_train == 0).sum(), (y_train == 1).sum()]
total_apres  = sum(counts_apres)

fig, ax = plt.subplots()
bars = ax.bar(['Non diabétique', 'Diabétique'], counts_apres, color=['steelblue', 'tomato'])
ax.set_title('Distribution après rééquilibrage SMOTE (50/50)')
ax.set_ylabel('Nombre de patients')
ax.yaxis.grid(False)
ax.set_axisbelow(False)

for bar, count in zip(bars, counts_apres):
    pct = count / total_apres * 100
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 500,
            f'{pct:.1f}%',
            ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```

## 16. Étape 8 : Normalisation

On applique un **StandardScaler** uniquement sur les variables continues et ordinales : BMI, MentHlth, PhysHlth, GenHlth et Age. Les variables binaires (0/1) restent en l'état car elles sont déjà sur la même échelle.

Le StandardScaler transforme chaque variable pour qu'elle ait une moyenne de 0 et un écart-type de 1. Cela évite que le modèle accorde plus d'importance à une variable simplement parce que ses valeurs sont numériquement plus grandes que les autres.

Cette méthode a été choisie car les réseaux de neurones convergent mieux avec des données centrées autour de 0. Le scaler est ajusté uniquement sur le train et appliqué ensuite sur val et test, ce qui évite tout data leakage : les données de validation et de test ne doivent jamais influencer les paramètres de normalisation.

Concrètement, chaque valeur est transformée selon la formule : **z = (valeur − moyenne) / écart-type**  
Par exemple, un BMI de 35 avec une moyenne de 28.6 et un écart-type de 6.7 donne z = (35 − 28.6) / 6.7 = +0.95, ce qui veut dire que cette personne a un BMI presque 1 écart-type au-dessus de la moyenne. Toutes les variables sont ainsi exprimées sur la même échelle, centrée en 0.

```python
# Étape 8 : Normalisation de toutes les variables
# On applique le StandardScaler sur toutes les colonnes d'un coup
# Fit uniquement sur X_train pour éviter tout data leakage

scaler = StandardScaler()

# Fit sur le train uniquement, puis transform sur les 3 splits
X_train = scaler.fit_transform(X_train)
X_val   = scaler.transform(X_val_raw)
X_test  = scaler.transform(X_test_raw)

# Vérification : les moyennes doivent être proches de 0 et les écarts-types proches de 1
print("Moyennes sur le train (doivent être ~0) :")
print(X_train.mean(axis=0).round(4))
print("\nÉcarts-types sur le train (doivent être ~1) :")
print(X_train.std(axis=0).round(4))

# Sauvegarde du scaler
scaler_path = PROCESSED_DIR / "scaler.joblib"
joblib.dump(scaler, scaler_path)
print(f"\nScaler sauvegardé : {scaler_path}")
```

## 18. Analyse RGPD & Éthique

Travailler avec des données de santé implique des responsabilités. Ce sprint a été l'occasion de prendre plusieurs décisions concrètes pour s'assurer que le projet respecte à la fois la loi et les principes éthiques.

**Les données source sont déjà anonymisées**

Le dataset BRFSS 2015 a été publié par les CDC après un travail d'anonymisation. Les participants ne sont identifiables par aucun moyen direct : pas de nom, pas d'adresse, pas de date de naissance précise. L'âge est regroupé en 13 tranches et les zones géographiques fines ont été retirées. On ne peut donc pas remonter à une personne en particulier à partir de ce dataset.

**Les variables discriminantes ont été retirées**

Trois variables ont été supprimées avant toute analyse : le sexe, le niveau d'études et le revenu. Ces variables ne sont pas des indicateurs médicaux du diabète et les conserver aurait pu amener le modèle à discriminer selon le genre, le niveau social ou les revenus. Un médecin ne diagnostique pas le diabète en fonction du diplôme ou du salaire de son patient, le modèle ne doit pas le faire non plus.

**Les doublons ont été supprimés**

Conserver des lignes en double aurait donné plus de poids à certains profils qu'à d'autres sans raison valable. La suppression des doublons est conforme au principe de minimisation des données prévu par le RGPD (art. 5.1.c).

**Le data leakage a été évité**

Le StandardScaler a été ajusté uniquement sur les données d'entraînement, puis appliqué sur la validation et le test. Cela garantit que le modèle ne bénéficie d'aucune information provenant des données qu'il ne devrait pas avoir vues.

**Ce modèle n'est pas un outil médical**

Il est important de rappeler que ce modèle est un outil d'aide à la décision et non un dispositif médical. Il ne peut pas et ne doit pas servir à diagnostiquer le diabète de façon automatisée sur un individu réel.

## 20. Glossaire

**Binarisation** : transformation d'une variable à plusieurs valeurs en une variable à deux valeurs seulement. Ici on a regroupé "pas de diabète" et "prédiabète" en une seule classe 0, et "diabète" en classe 1.

**Data leakage** : situation où des informations provenant des données de test ou de validation influencent l'entraînement du modèle. C'est une erreur grave qui donne l'illusion de bonnes performances mais qui ne se confirme pas en production.

**Déséquilibre des classes** : situation où une classe est bien plus représentée que l'autre dans le dataset. Ici 86% des individus n'ont pas de diabète contre 14% qui en ont.

**EDA (Exploratory Data Analysis)** : analyse exploratoire des données. Étape où on explore et visualise les données avant de construire un modèle pour mieux comprendre leur structure et leurs caractéristiques.

**Imbalance ratio** : rapport entre le nombre d'individus dans la classe majoritaire et la classe minoritaire. Ici 6.18, ce qui signifie qu'il y a 6.18 fois plus de non-diabétiques que de diabétiques.

**Normalisation** : transformation des valeurs d'une variable pour les ramener sur une échelle commune. Ici on utilise le StandardScaler qui centre chaque variable à une moyenne de 0 et un écart-type de 1.

**Overfitting** : situation où un modèle apprend trop bien les données d'entraînement au point de ne plus être capable de généraliser sur de nouvelles données. C'est pour détecter ce phénomène qu'on utilise un set de validation pendant l'entraînement.

**ROC AUC** : métrique d'évaluation qui mesure la capacité du modèle à distinguer les cas positifs des cas négatifs. Une valeur de 1 est parfaite, 0.5 équivaut à une prédiction aléatoire. Elle est plus adaptée que l'accuracy quand les classes sont déséquilibrées.

**Split train/val/test** : découpage du dataset en trois parties. Le train sert à l'entraînement, la validation à surveiller les performances pendant l'entraînement, et le test à l'évaluation finale du modèle.

**StandardScaler** : outil de normalisation qui transforme chaque variable pour qu'elle ait une moyenne de 0 et un écart-type de 1. Ajusté uniquement sur les données d'entraînement pour éviter tout data leakage.

**Stratification** : technique qui garantit que chaque partie du dataset (train, val, test) contient le même pourcentage de cas positifs que le dataset complet. Indispensable quand les classes sont déséquilibrées.

**SMOTE (Synthetic Minority Oversampling Technique)** : technique de rééquilibrage des classes qui génère des exemples synthétiques de la classe minoritaire en interpolant entre des points existants. Contrairement à une simple duplication de lignes, SMOTE crée de nouveaux exemples plausibles, ce qui enrichit réellement le jeu d'entraînement. Il est appliqué uniquement sur les données d'entraînement pour ne pas biaiser l'évaluation.
